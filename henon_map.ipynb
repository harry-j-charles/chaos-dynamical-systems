```python
import numpy as np
import matplotlib.pyplot as plt

a = 1.4
b = 0.3
num_iterations = 10000

x = np.zeros(num_iterations)
y = np.zeros(num_iterations)

x[0] = 0
y[0] = 0



for i in range(num_iterations - 1):
    x[i+1] = 1 - a * x[i]**2 + y[i]
    y[i+1] = b * x[i]

plt.figure(figsize=(10, 8))
plt.scatter(x, y, c="b", s = 0.5)
plt.title("Figure 1: Henon Attractor")
plt.xlabel("$x_n$")
plt.ylabel("$y_n$")
plt.show()

line_x = np.linspace(-0.5, 0.5, 500)
line_y = np.zeros(500)

plt.figure(figsize=(10, 8))


colors = ["red", "orange", "green", "purple", "blue"]

for i in range(5):

    plt.scatter(line_x, line_y, s=0.5, label=f"Iteration {i}", color=colors[i])

    new_x = 1 - a * line_x**2 + line_y
    new_y = b * line_x

    line_x = new_x
    line_y = new_y

plt.title("Figure 2: The Folding Mechanism at a=1.4")
plt.xlabel("$x_n$")
plt.ylabel("$y_n$")
plt.legend()
plt.show()



```
