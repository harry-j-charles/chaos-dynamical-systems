```python
import numpy as np
import matplotlib . pyplot as plt

r = 2.75


orbit = np . zeros (50)
orbit [0] = 0.5
for i in range (49) :
    orbit [ i +1] = r * orbit [ i ] * (1 - orbit [ i ])

plt.plot( range(50) , orbit , "o-", c = "blue", alpha = 0.5 )
plt.title("Logistic map orbit for r = 2.75")
plt.xlabel ("Iteration n")
plt.ylabel ("Xn")
plt.ylim(0,1)
plt.show ()
```
