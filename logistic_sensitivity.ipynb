```python
import numpy as np
import matplotlib . pyplot as plt

r = 3.75

orbit1 = np.zeros (80)
orbit1[0] = 0.5
for i in range (79) :
    orbit1[i +1] = r * orbit1[i] * (1 - orbit1[i])

orbit2 = np.zeros (80)
orbit2[0] = 0.5 + 1e-6
for i in range (79) :
    orbit2[i +1] = r * orbit2[i] * (1 - orbit2[i])


difference = np.abs(orbit1 - orbit2)

plt.plot(range(80), orbit1, "o-", c="blue", alpha=0.5, label="Orbit 1 ($x_0 = 0.5$)")
plt.plot(range(80), orbit2, "x--", c="red", alpha=0.5, label="Orbit 2 ($x_0 = 0.5 + 10^{-6}$)")
plt.title(f"Logistic Map Orbits (r = {r})")
plt.xlabel("Iteration n")
plt.ylabel("xn")
plt.ylim(0, 1)
plt.legend()
plt.show()


plt.plot(range(80), difference, "o-", c="purple")
plt.yscale("log")
plt.title("Absolute Difference between Orbits (Log Scale)")
plt.xlabel("Iteration n")
plt.ylabel("Difference |x1 - x2|")
plt.grid(True, which="both", linestyle="--", alpha=0.5)

plt.show()
```
